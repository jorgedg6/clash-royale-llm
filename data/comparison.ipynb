```python
import pandas as pd
import json
import numpy as np
from IPython.display import display
```

# Different samples

## All users

```python
user_cards = pd.read_csv("./player_cards.csv")
card_cols = user_cards.columns[1:]
user_cards['num_cards'] = (user_cards[card_cols] > 0).sum(axis=1)
user_cards
```

```python
user_stats = pd.read_csv("./player_stats.csv")
user_stats['win_rate'] = user_stats['wins'] / (user_stats['wins'] + user_stats['losses'])
user_stats = user_stats.merge(user_cards[['tag', 'num_cards']], on='tag', how='left')
user_stats
```

## Baseline users

```python
baseline_train = pd.read_csv("./train.csv")
baseline_test = pd.read_csv("./test.csv")
baseline_dataset = pd.concat([baseline_train, baseline_test], ignore_index=True)
```

```python
baseline_users = baseline_dataset["userID"].unique()
```

```python
baseline_users_stats = user_stats[user_stats["tag"].isin(baseline_users)]
baseline_users_stats
```

## LLM users

```python
unique_users = set()
path = "../results/generations_20251128_082528.jsonl" 

with open(path, 'r', encoding='utf-8') as f:
    for line in f:
        try:
            data = json.loads(line)
            if 'user_id' in data:
                unique_users.add("#" + data['user_id'])
                
        except json.JSONDecodeError:
            continue

print(f"Se encontraron {len(unique_users)} usuarios únicos.")
```

```python
llm_users_stats = user_stats[user_stats["tag"].isin(unique_users)]
llm_users_stats
```

# Comparison

```python
def get_column_statistics(df, column):
    mean = df[column].mean()
    median = df[column].median()
    std = df[column].std()

    print(f"[{column}] Media: {mean}, Mediana: {median}, Desviación: {std}")

    return mean, median, std
```

```python
dataframes = {
    "all": user_stats,
    "baseline": baseline_users_stats,
    "llm": llm_users_stats
}

column_names = ['win_rate', 'expLevel', 'num_cards']

summary = {}

for df_name, df in dataframes.items():
    summary[df_name] = {}
    for column in column_names:
        media, median, std = get_column_statistics(df, column)
        summary[df_name][column] = {
            "media": media,
            "median": median,
            "std": std
        }
```

```python
data = []
for group, vars_dict in summary.items():
    for variable, stats in vars_dict.items():
        for stat_name, value in stats.items():
            data.append({
                'Dataset': group,
                'Game Stat': variable,
                'Metric': stat_name,
                'Value': value
            })

df = pd.DataFrame(data)

tabla_final = df.pivot_table(index='Dataset', 
                             columns=['Game Stat', 'Metric'], 
                             values='Value')

tabla_final = tabla_final.reindex(['all', 'baseline', 'llm'])
```

```python
format = {
    'win_rate': '{:.1%}',
    'expLevel': '{:.1f}',
    'num_cards': '{:.1f}'
}

for column in column_names:
    sub_tabla = tabla_final[column]
    
    print(f"Estadísticas para: {column}")
    
    if column in format:
        styler = sub_tabla.style.format(format[column])
        display(styler)
    else:
        display(sub_tabla)
        
    print("\n")
```
